```python
import pandas as pd

archivos = {
    2018: "data/raw/h209.xlsx",
    2019: "data/raw/h216.xlsx",
    2020: "data/raw/h224.xlsx",
    2021: "data/raw/h233.xlsx",
    2022: "data/raw/h243.xlsx",
    2023: "data/raw/h251.xlsx",
}

for anio, path in archivos.items():
    df = pd.read_excel(path, nrows=1)
    yy = str(anio)[2:]
    cols = [c for c in df.columns if c.endswith(yy) or c.endswith(f"{yy}X") or c.endswith(f"{yy}F")]
    print(f"\n{anio} — columnas con sufijo {yy}:")
    print(cols)
```
